```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/processed_data.csv")

print(df.head())
```

```python
print("Shape of dataset:", df.shape)
print("\nColumns:")
print(df.columns)
```

```python
df['Original_Price'] = df['Price']
df.head()
```

```python
df['Adjusted_Price'] = df['Price']
df.head()
```

```python
df.loc[df['Demand Forecast'] > 120, 'Adjusted_Price'] *= 1.10
df.loc[df['Demand Forecast'] < 50, 'Adjusted_Price'] *= 0.95
```

```python
df['Date'] = pd.to_datetime(df['Date'])
print(df.dtypes)
```

```python
df['day_name'] = df['Date'].dt.day_name()
df['month'] = df['Date'].dt.month
df['is_weekend'] = df['Date'].dt.weekday >= 5

df[['Date', 'day_name', 'month', 'is_weekend']].head()
```

```python
df['original revenue'] = df['Price'] * df['Units Sold']
```

```python
df['original_revenue'] = df['Price'] * df['Demand Forecast']
```

```python
# Calculate original revenue
df['original_revenue'] = df['Price'] * df['Demand Forecast']

# Display result
df[['Price','Demand Forecast','original_revenue']].head()
```

```python
df['original_revenue'] = df['Price'] * df['Units Sold']
df[['Price', 'Units Sold', 'original_revenue']].head()
```

```python
df['adjusted_price'] = df['Price']
df[['Price', 'adjusted_price']].head()
```

```python
# Weekend price increase
df.loc[df['is_weekend'] == True, 'adjusted_price'] *= 1.10

# High-demand months price increase
df.loc[df['month'].isin([11, 12]), 'adjusted_price'] *= 1.12

# Low-demand months price decrease
df.loc[df['month'].isin([6, 7]), 'adjusted_price'] *= 0.95

df[['Date', 'Price', 'adjusted_price']].head(10)
```

```python
df['Original_Price'] = df['Price']
df['Adjusted_Price'] = df['Price']

df[['Price', 'Original_Price', 'Adjusted_Price']].head()
```

```python
df.loc[df['Demand Forecast'] > 120, 'Adjusted_Price'] *= 1.10
df.loc[df['Demand Forecast'] < 50, 'Adjusted_Price'] *= 0.95

df[['Demand Forecast', 'Price', 'Adjusted_Price']].head()
```

```python
df.loc[df['Seasonality'] == 'Summer', 'Adjusted_Price'] *= 1.08
df.loc[df['Seasonality'] == 'Winter', 'Adjusted_Price'] *= 1.05

df[['Seasonality', 'Price', 'Adjusted_Price']].head()
```

```python
df.loc[df['Holiday/Promotion'] == 1, 'Adjusted_Price'] *= 1.07

df[['Holiday/Promotion', 'Price', 'Adjusted_Price']].head()
```

```python
df.loc[df['Price'] < df['Competitor Pricing'], 'Adjusted_Price'] *= 1.05
df.loc[df['Price'] > df['Competitor Pricing'] + 10, 'Adjusted_Price'] *= 0.97

df[['Price', 'Competitor Pricing', 'Adjusted_Price']].head()
```

```python
df['Adjusted_Price'] = df['Adjusted_Price'].round(2)

df[['Price', 'Adjusted_Price']].head()
```

```python
df['Original_Revenue'] = df['Original_Price'] * df['Demand Forecast']

df[['Original_Price', 'Demand Forecast', 'Original_Revenue']].head()
```

```python
df['New_Revenue'] = df['Adjusted_Price'] * df['Demand Forecast']

df[['Adjusted_Price', 'Demand Forecast', 'New_Revenue']].head()
```

```python
df[['Price', 'Adjusted_Price', 'Demand Forecast', 'Original_Revenue', 'New_Revenue']].head(10)
```

```python
total_original_revenue = df['Original_Revenue'].sum()
total_new_revenue = df['New_Revenue'].sum()

print("Total Original Revenue:", round(total_original_revenue, 2))
print("Total New Revenue:", round(total_new_revenue, 2))
```

```python
revenue_lift = ((total_new_revenue - total_original_revenue) / total_original_revenue) * 100
print("Revenue Lift (%):", round(revenue_lift, 2))
```

```python
print("========== FINAL RESULTS ==========")
print("Original Revenue :", round(total_original_revenue, 2))
print("New Revenue      :", round(total_new_revenue, 2))
print("Revenue Lift %   :", round(revenue_lift, 2))
```

```python
df.to_csv('/content/pricing_engine_output.csv', index=False)
print("File saved successfully!")
```
